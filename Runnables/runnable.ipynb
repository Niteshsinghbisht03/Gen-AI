```python
import random

class NakliLLM:
    def __init__(self):
        print('LLM created')
        
    def predict(self,prompt):
        response_list = [
            'Hii me hu first',
            'Hii me hu second',
            'Hii me hu third',
        ]
        return {'response':random.choice(response_list)}
```

```python
llm = NakliLLM()
```

```python
llm.predict('hello which number')
```

```python
class NakliPromptTemplate:
    def __init__(self,template,input_variables):
        self.template = template
        self.input_variables = input_variables
        
    def format(self,input_dict):
        return self.template.format(**input_dict)
        
        
```

```python
template = NakliPromptTemplate(
    template='write a {length} poem about {topic}',
    input_variables=['topic','length']
)

prompt = template.format(
    {
        'length':'short',
        'topic':'LLMs'
        }
    )
```

```python
llm.predict(prompt)
```

Nakli LLM Chain

```python
class NakliLLMChain:
    def __init__(self,llm,template):
        self.llm = llm
        self.template = template
        
    def run(self,input_dict):
        final_prompt = self.template.format(input_dict)
        result = self.llm.predict(final_prompt)
        return result['response']
```

```python
llm = NakliLLM()
```

```python
template = NakliPromptTemplate(
    template='write a {length} poem about {topic}',
    input_variables=['topic','length']
)
```

```python
chain = NakliLLMChain(llm,template)

chain.run({'length':'short','topic':'LLMs'})
```

Asli Runnables

```python
from abc import ABC , abstractmethod

class Runnable(ABC):
    
    @abstractmethod
    def invoke(input_data):
        pass
```

```python
import random

class NakliRunnableLLM(Runnable):
    def __init__(self):
        print('Runnable created')

    def invoke(self,prompt):
        response_list = [
            'Hii me hu first',
            'Hii me hu second',
            'Hii me hu third',
        ]
        return {'response':random.choice(response_list)}
    
    def predict(self,prompt):
        response_list = [
            'Hii me hu first',
            'Hii me hu second',
            'Hii me hu third',
        ]
        return {'response':random.choice(response_list)}
```

```python
llm = NakliRunnableLLM()
llm.invoke('hello ji')
```

```python
class NakliPromptTemplate(Runnable):
    def __init__(self,template,input_variables):
        self.template = template
        self.input_variables = input_variables
       
       
    def invoke(self,input_dict):
        return self.template.format(**input_dict) 
    
    def format(self,input_dict):
        return self.template.format(**input_dict)
        
```

```python
prompt = NakliPromptTemplate(
    template='hii ji {topic}',
    input_variables=['topic']
)
prompt.invoke({'topic':'acricket'})
```

```python
class NakliStrParser(Runnable):
    def __init__(self):
        pass
    
    def invoke(self,input_data):
        return input_data['response']
```

```python
strParser = NakliStrParser()
```

```python
class RunnableConnector(Runnable):
      def __init__(self,runnable_list):
          self.runnable_list= runnable_list
          
      def invoke(self,input_data):
          for runnable in self.runnable_list:
              input_data = runnable.invoke(input_data)
          return input_data
```

```python
chain = RunnableConnector([prompt,llm,strParser])
```

```python
chain.invoke({'topic':'india'})
```

```python
template1 = NakliPromptTemplate(
    template='write a joke about {topic}',
    input_variables=['topic']
)

template2 = NakliPromptTemplate(
    template='Explain the following \n {response}',
    input_variables=['response']
)
```

```python
llm = NakliRunnableLLM()
```

```python
chain1 = RunnableConnector([template1,llm])
```

```python
chain2 = RunnableConnector([template2,llm,strParser])
```

```python
final_chain = RunnableConnector([chain1,chain2])
final_chain.invoke({'topic':'cricket'})
```
