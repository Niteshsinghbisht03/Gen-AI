```python
from langchain.schema import Document
from langchain_openai import OpenAIEmbeddings
from langchain.vectorstores import Chroma
from langchain.embeddings import HuggingFaceEmbeddings
```

```python
doc1 = Document(
    page_content="Virat Kohli is one of the most consistent and aggressive batsmen in modern cricket. Known for his passion and intensity, he has been a pillar for RCB since the beginning of the IPL. He is also a former captain of both India and RCB and holds the record for the most runs in IPL history.",
    metadata={"team":"Royal Challengers Bangalore (RCB)"}
)

doc2 = Document(
    page_content="MS Dhoni, often called Captain Cool, is a legendary cricketer known for his calm mind and brilliant finishing skills. He has led CSK to multiple IPL titles and is admired for his sharp captaincy and wicket-keeping. Even in the later years of his career, he continues to be a fan favorite.",
    metadata={"team":"Chennai Super Kings (CSK)"}
)

doc3 = Document(
    page_content="Rohit Sharma is a stylish and explosive opening batsman who has led Mumbai Indians to five IPL titles as captain. He is known for his effortless batting technique and leadership qualities. Rohit is also the captain of the Indian national team across all formats.",
    metadata={"team":"Mumbai Indians (MI)"}
)

doc4 = Document(
    page_content="Shubman Gill is a young and talented top-order batsman who has impressed everyone with his elegant stroke play and consistency. He was one of the top scorers in IPL 2023 and played a key role in Gujarat Titans' performance. Gill is seen as a future star of Indian cricket.",
    metadata={"team":"Gujarat Titans (GT)"}
)

doc5 = Document(
    page_content="KL Rahul is a stylish right-handed batsman and wicketkeeper who is known for his calm demeanor and ability to anchor an innings. He is the captain of LSG and has consistently been among the top run-scorers in the IPL over the past few seasons.",
    metadata={"team":"Lucknow Super Giants (LSG)"}
)
```

```python
docs=[doc1,doc2,doc3,doc4,doc5]
```

```python
vector_store = Chroma(
    embedding_function = HuggingFaceEmbeddings(model_name='sentence-transformers/all-MiniLm-L6-v2'),
    persist_directory='my_chroma_db',
    collection_name = 'sample',
)
```

```python
vector_store.add_documents(docs)
```

```python
vector_store.get(include = ['embeddings','documents','metadatas'])
```

```python
vector_store.similarity_search(
    query= 'who is the most young amoung these?',
    k=2
)
```

```python
vector_store.similarity_search_with_score(
    query= 'who is the most young amoung these?',
    k=2
)
```

```python
vector_store.similarity_search_with_score(
    query= '',
    filter = {'team':'Chennai Super Kings (CSK)'}
)
```
